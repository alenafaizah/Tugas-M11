```python
#Soal 3.1

def hitung_luas_persegi(sisi):
    luas=sisi*sisi
    print(f'Luas persegi adalah: {luas}')
    return luas

def rata_rata(nilai1, nilai2, nilai3):
    rata_rata=(nilai1+nilai2+nilai3)/3
    return rata_rata

hasil1=hitung_luas_persegi(5)
hasil2=hitung_luas_persegi(10)
hasil3=hitung_luas_persegi(15)

print(f'Rata rata 3 persegi: {rata_rata(hasil1,hasil2,hasil3)}')
```

```python
#Soal 5

def konversi_nilai(nilai_angka):
    if (nilai_angka>=86):
        return "A"
    elif (nilai_angka>=76 and nilai_angka<86):
        return "AB"
    elif (nilai_angka>=66 and nilai_angka<76):
        return "B"
    else:
        return "C"

nilai=int(input("Masukkan nilai: "))

indeks=konversi_nilai(nilai)
print(f'Nilai huruf Anda adalah: {indeks}')
```
